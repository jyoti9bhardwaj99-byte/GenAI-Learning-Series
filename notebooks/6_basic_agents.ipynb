```python
from langchain_core.tools import tool
```

```python
@tool
def add_number(a:int, b:int):
    """ 
        it will return the sum of two number

        Args:
            a : Number one
            b : Number Two
            
    """

    return a + b


@tool
def multiply_number(a:int, b:int):
    """ 
        it will return the product of two number

        Args:
            a : Number one
            b : Number Two
            
    """

    return a * b
```

```python
add_number.invoke({"a":12, "b":3})
```

```python
from langchain_groq import ChatGroq
from langchain.agents import create_agent
```

```python
llm = ChatGroq(model="llama-3.3-70b-versatile")
```

```python
agent = create_agent(
    model=llm,
    tools=[add_number],
    system_prompt="You are a maths teacher and you always use tool for calculation"
)
```

```python
res = agent.invoke({"messages":[{"role":"user", "content": "Wht is 2 + 30 ?"}]})
```
