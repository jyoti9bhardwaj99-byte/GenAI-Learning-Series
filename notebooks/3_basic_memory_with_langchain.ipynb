```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_groq import ChatGroq
llm = ChatGroq(model="llama-3.3-70b-versatile")
```

```python
prompts = [
    {"role":"user","content":"Hello, My name is jyoti bhardwaj"},
    {"role":"ai","content":"Hello Jyoti Bhardwaj, it's nice to meet you.Is there something I can help you with or would you like to chat?"},
    {"role":"user","content":"What is my name?"} 
]

res =  llm.invoke(prompts)
```

```python
res.content
```

#### Manage History

```python
history = []
while True:
    query = input("user: ")
    if query.lower() in ["exit", "quit", "bye"]:
        print("Good Bye👋")
        break

    history.append({"role":"user", "content": query})
    print("user:", query)


    res = llm.invoke(history)
    history.append({"role":"ai", "content": res.content})
    print("AI:", res.content,  "\n")
```
