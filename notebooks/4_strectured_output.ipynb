#### Solution is " Structured Output

```python
from pydantic import BaseModel, Field

class ResponseStrecture(BaseModel):
    name:str = Field(description="Complete Name")
    email:str = Field(description="Email Address")
    age:int = Field(description="Age")

strectured_llm = llm.with_structured_output(ResponseStrecture)    
```

```python
res = strectured_llm.invoke(f"Please give me only name,age and email from this text:  {text}")

res
```

```python
res.name
```

```python
class Movies(BaseModel):
    title:str = Field(description="Movie Title.")
    year:int = Field(description="Movie Released Year.")

movie_llm = llm.with_structured_output(Movies)    
```

```python
movie_llm.invoke("Give me any resent Comady movie and released year of movie")
```

#### Now if we want list of movies

```python
from typing import List
```

```python
class Movies(BaseModel):
    title:str = Field(description="Movie Title.")
    year:int = Field(description="Movie Released Year.")

class AllMovies(BaseModel):
    movies:List[Movies]  

movie_llm = llm.with_structured_output(AllMovies)    
```

```python
movie_llm.invoke("Give me 3 resent Comady movie and released year of movie")
```
