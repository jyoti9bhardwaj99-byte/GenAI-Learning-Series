#### Prompts Templates- Dynamic Prompts

```python
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
llm = ChatGroq(model="llama-3.3-70b-versatile")
out = StrOutputParser()
```

```python
#### My self chain node

def transform_case(result:str):
    return result.upper()
```

```python
prompts = ChatPromptTemplate([
    {"role":"system", "content":"You are a translator and translate input in {language}"},
    {"role":"user", "content":"{query}"},
])
prompts

```

```python
#### Now we make it standard chaning 
```

```python
res = llm.invoke(prompts.invoke({"language": "hindi","query": "I love hindi"}))
res.content
```

```python
#### Know this is standard method to covert into chains
```

```python
chains = prompts | llm |out |transform_case
```

```python
res = chains.invoke({"language": "hinglish","query": "I love langchain"})
print(res)
```
