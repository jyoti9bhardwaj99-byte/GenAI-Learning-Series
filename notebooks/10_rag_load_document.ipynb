#### Text Documents Loader


```python
from langchain_community.document_loaders import TextLoader
loader = TextLoader("../data/speech.txt", encoding="utf-8")
texts = loader.load()
len(texts) # print lenth of the content

oneDoc = texts[0]
print(oneDoc.page_content) # print  all the content of the page
```

#### PDF Document Loader

```python
from langchain_community.document_loaders import PyPDFLoader
loader= PyPDFLoader("../data/medical_report.pdf")
docs = loader.load()
len(docs)    # print length of the content


```
