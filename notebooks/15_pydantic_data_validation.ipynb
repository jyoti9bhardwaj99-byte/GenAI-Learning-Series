```python
from pydantic import BaseModel, Field
from typing import List, Annotated
```

```python
class User(BaseModel):
    name:str 
    age:int
```

```python
user = User(name = "jyoti", age = 22)
user
```

```python

#### Field Validation

class Product(BaseModel):
    title:str = Field(max_length=10) 
    des:str = Field(min_length=20)
    price:int = Field(gt=0, le=100)
```

```python
product = Product(title="SamsungS23", des="This is a very amazing smart phone from S23 series.", price=100)
product2 = Product(title="Oppo S21", des="This is a very amazing smart phone from S23 series.", price=100)
```

```python
product
```

#### Here we connect object of different class in another class

```python
class Order(BaseModel):
    id:str = Field(min_length=6 , max_length=12)
    product:Product
```

```python
order = Order(id = "guyguv76", product=product)
```

```python
order
order.product.price
```

```python
class Carts(BaseModel):
    id:str = Field(min_length=6 , max_length=12)
    items: List[Product]
    comment:str  = Field(
        default="Nice Product" 
    )
```

```python
cart = Carts(id= "uydeghbkj", items=[product, product2])
```

```python
cart
```

```python
for product in cart.items:
    print(product.title)
```

```python
# Annotated

CommonValidator = Annotated[List[Product], "Field(default=[])"]
```

```python
class Carts(BaseModel):
    id:str = Field(min_length=6 , max_length=12)
    items: CommonValidator
    comment:str  = Field(
        default="Nice Product" 
    )
```

```python
class Order(BaseModel):
    product :CommonValidator
```

```python
ord = Order(product=[product,product2])
```
