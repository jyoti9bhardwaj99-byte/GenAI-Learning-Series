#### Importing All The Libries

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_ollama import OllamaEmbeddings
from langchain_chroma import Chroma
from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate
```

#### Loading The Document

```python
loader = PyPDFLoader("../data/data_science_syllabus.pdf")
docs = loader.load()
len(docs)
docs[0].page_content
```

#### Splitting The Data|

```python
splitter = RecursiveCharacterTextSplitter(chunk_size = 600, chunk_overlap = 100)
splitted_data = splitter.split_documents(docs)
splitted_data
len(splitted_data)
```

#### Converting In Embedding And Storing In Chroma

```python
embeddings = OllamaEmbeddings(model="nomic-embed-text")
```

```python
vector_store = Chroma.from_documents(
    documents=splitted_data,
    embedding=embeddings
)
    
```

```python
query = "Machine Learning and Data Science Content"
data = vector_store.similarity_search(query=query)
```

#### Genterating Context From Data For LLM

```python
context = ""
for doc in data:
    context += doc.page_content + "\n"

```

```python
llm = ChatGroq(model="llama-3.3-70b-versatile")
```

```python
#### Chain = Content_generate  / propt  /  llm  /  straparser
```

```python
def get_context(query:str):
    data = vector_store.similarity_search(query=query)
    context = ""
    for doc in data:
        context += doc.page_content + "\n"

    return {
        "context":context,
        "question":query
    }    


```

```python
prompt = PromptTemplate.from_template ("""
    You are a helpful assistant and provide answerd based on the context for user question. 
    and if you don't know the answer you can say that 'I dont know.'
    context: {context}
    question: {question}
    """)
```

```python
rag_chain = get_context | prompt | llm
```

```python
res = rag_chain.invoke("Give me complete content for Gen Ai module") # we can ask any question from the uploaded pdf 
```
